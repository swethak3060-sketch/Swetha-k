```python
password=input("Enter the password")
length=len(password)
has_letter=False
has_digit=False
has_special=False
for ch in password:
    if ch.isalpha():
        has_letter=True
    elif ch.isdigit():
        has_digit=True
    else:
        has_special=True

if length>=8 and has_letter and has_digit and has_special:
    print("Strong password")
elif length>=8 and has_letter and has_digit and not has_special:
    print("Moderate password")
    print("Add a special character")
elif length>=8 and has_letter and has_special and not has_digit:
    print("Strong-Moderate password")
elif length>=8 and has_letter and not has_digit and not has_special:
    print("week password")
    print("Add a special character and digits")
else:
    print("very week password")
    print("Atleat 8 character,one digit and one special character")
```
