```python
# while True:
    salary = float(input("Enter base salary: "))
    if salary <= 0:
        print("Salary must be positive. Try again.")
        continue

    exp = int(input("Enter years of experience: "))
    if exp < 0:
        print("Experience cannot be negative. Try again.")
        continue

    if exp >= 10:
        salary += salary * 0.15
    elif 5 <= exp <= 9:
        salary += salary * 0.10
    elif 2 <= exp <= 4:
        salary += salary * 0.05
    else:
        pass  

    rating = int(input("Enter performance rating (1-5): "))
    if rating < 1 or rating > 5:
        print("Invalid rating. Must be between 1 and 5.")
        continue
 
    if rating == 5:
        salary += salary * 0.10
    elif rating == 4:
        salary += salary * 0.05
    else:
        pass  # rating 1-3 → no increment

    print(f"Employee's new salary: {round(salary, 2)}")

    more = input("Do you want to enter another employee? (yes/no): ").lower()
    if more != 'yes':
        print("Salary processing completed.")
        break
```
