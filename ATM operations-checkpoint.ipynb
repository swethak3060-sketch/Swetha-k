```python
balance = 1000
pin = 1234
for attempts in range(3):
    user_pin = int(input("Enter the PIN: "))
    if user_pin == pin:
        print("Login successful")
        success=True
        break
    else:
        print("Wrong PIN")
else:
    print("Account locked")
if success:
    while True:
        print("SELECT")
        print("\n1: Check Balance")
        print("\n2: Deposit Money")
        print("\n3: Withdraw Money")
        print("\n4: Exit")
        
        ch = int(input("Enter your choice: "))

        if ch == 1:
            print("BALANCE:", balance)

        elif ch == 2:
            amt = int(input("Enter the amount you would like to deposit: "))
            if amt <0 or amt==0:
                print("The amount you have entered is Invalid")
            else:
                balance += amt
                print("Your current balance:", balance)

        elif ch == 3:
            amt = int(input("Enter the amount you would like to withdraw: "))
            if amt > balance:
                print("Insufficient funds")
            else:
                balance -= amt
                print("Your current balance:", balance)

        elif ch == 4:
            print("Thank you for using ATM")
            break

        else:
            print("Invalid choice")
```
