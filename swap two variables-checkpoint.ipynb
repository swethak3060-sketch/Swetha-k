```python
# k swetha 2025peccs535

# Define two variables
x = 5
y = 10

# Print values before swapping
print("Before swapping:")
print("x =", x, ", y =", y)

# Use a third variable to swap
temp = x
x = y
y = temp

# Print values after swapping
print("After swapping:")
print("x =", x, ", y =", y)
```
