```python
"""Basic calculator using iinput()function and f-striings to format"""
print("BASIC CALCULATOR")
x=int(input("Enter the value of x"))
y=int(input("Enter the value of y"))
#Addition
add=x+y
#Subtraction
sub=x-y
#Multiplication
mul=x*y
#Division
div=x/y
#Floor division
flr=x//y
#Exponentiation
exp=x**y
print(f"add={add},sub={sub},mul={mul},div={div},flr={flr},exp={exp}")
```
